## Part 1: Covariance and Correlation

```python
# 1.1 Import the libraries and load the data without missing values as `df`.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("penguins.csv").dropna()
```

```python
# 1.2 Calculate the covariance of flipper length and body mass yourself (divide by n).
x = df["flipper_length_mm"]
y = df["body_mass_g"]
cov = ((x - x.mean()) * (y - y.mean())).sum() / len(x)
print(cov)
```

```python
# 1.3 Calculate the correlation r = cov / (SD(x) × SD(y)). Use `np.std()`, which also divides by n. Check with `x.corr(y)`.
r = cov / (np.std(x) * np.std(y))
print(round(r,4), round(x.corr(y), 4))
```

```python
# 1.4 Draw a scatter plot of flipper length against body mass. Put r in the title.
plt.scatter(x, y, s=10)
plt.xlabel("Flipper length (mm)")
plt.ylabel("Body mass (g)")
plt.title(f"r = {r:.2f}")
plt.show()
```
